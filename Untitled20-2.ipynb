```python
import requests
import pandas as pd
import time

# Polygon.io API Key
API_KEY = "oyx5jTDifa9zxj_duV6zUu8fGh65RXtv"

# Function to fetch live EUR/USD data
def fetch_live_forex_data():
    url = f"https://api.polygon.io/v2/aggs/ticker/C:EURUSD/range/1/minute/{pd.Timestamp.now().date()}/{pd.Timestamp.now().date()}?adjusted=true&sort=asc&limit=50000&apiKey={API_KEY}"
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json().get("results", [])
        df = pd.DataFrame(data)
        df["timestamp"] = pd.to_datetime(df["t"], unit="ms")
        df.rename(columns={"o": "open", "h": "high", "l": "low", "c": "close", "v": "volume"}, inplace=True)
        df = df[["timestamp", "open", "high", "low", "close", "volume"]]
        return df
    else:
        print(f"Error fetching data: {response.status_code}")
        return None
```

```python
# ATR Calculation
def calculate_atr(data, period=14):
    data["tr"] = data[["high", "low", "close"]].apply(lambda x: max(x[0] - x[1], abs(x[0] - x[2]), abs(x[1] - x[2])), axis=1)
    data["atr"] = data["tr"].rolling(window=period).mean()

# CRT Strategy
def crt_strategy(data, atr_sl=True, atr_multiplier=1.5):
    trades = []
    calculate_atr(data)  # Calculate ATR

    for i in range(2, len(data)):
        left_candle = data.iloc[i - 2]
        middle_candle = data.iloc[i - 1]
        right_candle = data.iloc[i]

        # 🔻 Sell Setup
        if middle_candle["high"] > left_candle["high"] and middle_candle["close"] < left_candle["high"]:
            sl = right_candle["open"] + (atr_multiplier * middle_candle["atr"]) if atr_sl else left_candle["high"] + 0.5
            tp = left_candle["low"]
            trades.append({
                "type": "SELL",
                "entry": right_candle["open"],
                "stop_loss": sl,
                "take_profit": tp,
                "timestamp": right_candle["timestamp"]
            })

        # 🔹 Buy Setup
        elif middle_candle["low"] < left_candle["low"] and middle_candle["close"] > left_candle["low"]:
            sl = right_candle["open"] - (atr_multiplier * middle_candle["atr"]) if atr_sl else left_candle["low"] - 0.5
            tp = left_candle["high"]
            trades.append({
                "type": "BUY",
                "entry": right_candle["open"],
                "stop_loss": sl,
                "take_profit": tp,
                "timestamp": right_candle["timestamp"]
            })

    return trades
```

```python
import time

def simulate_live_forward_testing(atr_sl=True, interval=60, iterations=10):
    balance = 10000
    risk_per_trade = 0.0025  # 0.25% risk per trade
    all_trades = []

    for _ in range(iterations):  # Fetch live data for a limited number of iterations
        print(f"\nFetching live data at {pd.Timestamp.now()}")
        live_data = fetch_live_forex_data()

        if live_data is None or live_data.empty:
            print("No data fetched. Skipping iteration.")
            time.sleep(interval)
            continue

        trades = crt_strategy(live_data, atr_sl=atr_sl)
        for trade in trades:
            trade_risk = balance * risk_per_trade
            sl_distance = abs(trade["entry"] - trade["stop_loss"])
            position_size = trade_risk / sl_distance

            if trade["type"] == "BUY":
                if trade["take_profit"] > trade["entry"]:  # Simulate a win
                    balance += trade_risk
                    trade["outcome"] = "Win"
                else:  # Simulate a loss
                    balance -= trade_risk
                    trade["outcome"] = "Loss"
            elif trade["type"] == "SELL":
                if trade["take_profit"] < trade["entry"]:  # Simulate a win
                    balance += trade_risk
                    trade["outcome"] = "Win"
                else:  # Simulate a loss
                    balance -= trade_risk
                    trade["outcome"] = "Loss"

            trade["balance"] = balance
            all_trades.append(trade)

        print(f"Iteration completed. Current balance: ${balance:.2f}")
        time.sleep(interval)  # Wait for the next interval

    # Convert to DataFrame
    trades_df = pd.DataFrame(all_trades)
    return trades_df
```

```python
# Run the live forward test
live_test_results = simulate_live_forward_testing(atr_sl=True, interval=60, iterations=5)

# Save results to CSV
live_test_results.to_csv("Live_Forward_Test_Results.csv", index=False)
print("\nLive forward test results saved to 'Live_Forward_Test_Results.csv'")
```

```python
import matplotlib.pyplot as plt

# Plot balance over time
plt.figure(figsize=(12, 6))
plt.plot(live_test_results["timestamp"], live_test_results["balance"], label="Account Balance")
plt.title("Live Forward Test Balance Over Time")
plt.xlabel("Time")
plt.ylabel("Balance ($)")
plt.legend()
plt.grid()
plt.show()

# Plot win/loss distribution
plt.figure(figsize=(8, 4))
live_test_results["outcome"].value_counts().plot(kind="bar", color=["green", "red"])
plt.title("Win/Loss Distribution (Live Forward Test)")
plt.xlabel("Outcome")
plt.ylabel("Number of Trades")
plt.grid(axis="y")
plt.show()
```
